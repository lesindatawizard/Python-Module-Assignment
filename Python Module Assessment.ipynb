```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('module_project_data.csv')
```

```python
df
```

```python
# Exercise 1: Preprocessing
```

```python
# Step 1:  Correct the data in the "height" column by replacing it with random numbers between 150 and 180.
```

```python
df['Height'] = np.random.randint(150, 181)
```

```python
df.head()
```

```python
print("Missing values:\n", df.isnull().sum())
```

```python
# Ensure data consistency and integrity before proceeding with analysis.
```

```python
# Replacing the missing value of salary with average of all the values and also removing college column due to huge no. of null value
```

```python
median_salary = df['Salary'].median()
df['Salary'].fillna(median_salary, inplace=True)
```

```python
df
```

```python
df = df.drop('College', axis=1)
df
```

```python
print("Missing values:\n", df.isnull().sum())
```

```python
# Exercise 2: Analysis Tasks
```

```python
# Determine the distribution of employees across each team and calculate the percentage split relative to the total number of employees.
```

```python
team_dist = df['Team'].value_counts()
team_percent = (team_dist / len(df)) * 100
```

```python
team_dist
```

```python
team_percent
```

```python
plt.figure(figsize=(12,6))
ax = team_dist.plot(kind='bar', color='skyblue')

for i, v in enumerate(team_dist):
    ax.text(i, v + 0.2, str(v), ha='center', va='bottom', fontsize=10)

max_count = team_dist.max()
yticks = np.arange(0, max_count + 2, 2)
plt.yticks(yticks, fontsize=10)

plt.title('Employee Distribution by Team', fontsize=14, pad=20)
plt.xlabel('Team', fontsize=12)
plt.ylabel('Number of Employees', fontsize=12)
plt.xticks(rotation=45, fontsize=10, ha='right')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

plt.figure(figsize=(14,6))
team_percent.plot(kind='pie', autopct='%1.1f%%', 
                 textprops={'fontsize': 6})  # Percentage label size
plt.title('Percentage Distribution of Employees by Team', 
          fontsize=14, pad=20)  # Increased title size
plt.ylabel('', fontsize=12)  # Y-label (empty but sets size if used)
plt.tight_layout()
plt.show()
```

```python
# Segregate employees based on their positions within the company.
```

```python
position_dist = df['Position'].value_counts()
```

```python
position_dist
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
bars = position_dist.plot(kind='barh', color='lightgreen', ax=ax)

for i, v in enumerate(position_dist):
    ax.text(v / 2, i, str(v), color='black', va='center', ha='center', fontsize=10)

ax.set_title('Employee Distribution by Position')
ax.set_xlabel('Number of Employees')
ax.set_ylabel('Position')

plt.tight_layout()
plt.show()
```

```python
# Identify the predominant age group among employees.
```

```python
age_bins = [18, 25, 30, 35, 40, 45]
age_labels = ['18-25', '26-30', '31-35', '36-40', '41-45']
df['Age_Group'] = pd.cut(df['Age'], bins=age_bins, labels=age_labels)
age_group_dist = df['Age_Group'].value_counts()
```

```python
age_group_dist
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
bars = age_group_dist.plot(kind='bar', color='salmon', ax=ax)

for i, v in enumerate(age_group_dist):
    ax.text(i, v + 0.2, str(v), ha='center', va='bottom', fontsize=10)

ax.set_title('Employee Distribution by Age Group')
ax.set_xlabel('Age Group')
ax.set_ylabel('Number of Employees')

plt.tight_layout()
plt.show()
```

```python
# Discover which team and position have the highest salary expenditure.
```

```python
team_salary = df.groupby('Team')['Salary'].sum().sort_values(ascending=False)
position_salary = df.groupby('Position')['Salary'].sum().sort_values(ascending=False)
```

```python
position_salary
```

```python
team_salary
```

```python
fig, ax = plt.subplots(figsize=(10, 8))
bars = team_salary.plot(kind='bar', color='purple', ax=ax)

max_salary = team_salary.max()
yticks = np.arange(0, max_salary + max_salary * 0.1, max_salary * 0.1)
ax.set_yticks(yticks)
ax.tick_params(axis='y', labelsize=10)

ax.set_title('Total Salary Expenditure by Team', fontsize=14, pad=20)
ax.set_xlabel('Team', fontsize=12)
ax.set_ylabel('Total Salary', fontsize=12)
ax.set_xticklabels(team_salary.index, rotation=45, ha='right', fontsize=10)

ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(10, 6))
bars = position_salary.plot(kind='bar', color='orange', ax=ax)

max_salary = position_salary.max()
yticks = np.arange(0, max_salary + max_salary * 0.1, max_salary * 0.1)
ax.set_yticks(yticks)
ax.tick_params(axis='y', labelsize=10)

ax.set_title('Total Salary Expenditure by Position', fontsize=14, pad=20)
ax.set_xlabel('Position', fontsize=12)
ax.set_ylabel('Total Salary', fontsize=12)
ax.set_xticklabels(position_salary.index, rotation=45, ha='right', fontsize=10)

ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Investigate if there's any correlation between age and salary, and represent it visually.
```

```python
plt.figure(figsize=(10,6))
sns.regplot(x='Age', y='Salary', data=df, scatter_kws={'alpha':0.3})
plt.title('Correlation between Age and Salary')
plt.xlabel('Age')
plt.ylabel('Salary ($)')
plt.show()
```

```python
correlation = df['Age'].corr(df['Salary'])
print(f"Correlation between Age and Salary: {correlation:.2f}")
```
